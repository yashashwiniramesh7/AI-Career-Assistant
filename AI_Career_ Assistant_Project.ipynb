```python
# Load the small English model for processing
nlp = spacy.load("en_core_web_sm")
```

```python
def ai_career_agent(resume_text, job_description):
    # Step 4: Input Processing
    resume_doc = nlp(resume_text.lower())
    job_doc = nlp(job_description.lower())

    # Extracting keywords (Logic-based reasoning)
    resume_keywords = [token.text for token in resume_doc if token.pos_ in ['NOUN', 'PROPN']]
    job_keywords = [token.text for token in job_doc if token.pos_ in ['NOUN', 'PROPN']]

    # Identify missing skills by comparing the sets
    missing_skills = set(job_keywords) - set(resume_keywords)

    # Calculate a Match Score percentage
    match_score = len(set(resume_keywords) & set(job_keywords)) / len(set(job_keywords)) * 100

    # Output Generation
    print("--- AI Career Agent Analysis ---")
    print(f"Match Accuracy: {match_score:.2f}%")
    print(f"Top Missing Skills: {list(missing_skills)[:5]}")

# --- Step 4: Demonstrate Functionality (Interactivity) ---
print("Welcome to the AI Career Assistant!")
user_resume = input("Paste your Resume text here: ")
user_job = input("Paste the Job Description here: ")

ai_career_agent(user_resume, user_job)
```
